```python
from random import random

from langchain_core.runnables import RunnableLambda

# A RunnableSequence constructed using the `|` operator
sequence = RunnableLambda(lambda x: x + 1) | RunnableLambda(lambda x: x * 2)
sequence.invoke(1)  # 4
sequence.batch([1, 2, 3])  # [4, 6, 8]


# A sequence that contains a RunnableParallel constructed using a dict literal
sequence = RunnableLambda(lambda x: x + 1) | {
    "mul_2": RunnableLambda(lambda x: x * 2),
    "mul_5": RunnableLambda(lambda x: x * 5),
}
sequence.invoke(1)  # {'mul_2': 4, 'mul_5': 10}
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda
import asyncio

prompt = ChatPromptTemplate.from_template(
    "Write in short about {topic}."
)

parser = StrOutputParser()
```

```python
from langchain_ollama import ChatOllama

model = ChatOllama(
    model="gemma3:4b",
    temperature=1.9,
    top_p=0.95,
    top_k=50
)

chain = prompt | model | parser

```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=1.9,
    top_p=0.95
)

chain = prompt | model | parser
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",
    temperature=1.9,
    top_p=0.95
)

chain = prompt | model | parser
```

```python
from langchain_anthropic import ChatAnthropic

model = ChatAnthropic(
    model="claude-3-haiku-20240307",
    temperature=1.9,
    top_p=0.95
)

chain = prompt | model | parser
```

```python
from langchain_openai import AzureChatOpenAI

model = AzureChatOpenAI(
    azure_deployment="gpt-4o-mini",
    api_version="2024-02-01",
    temperature=1.9,
    top_p=0.95
)

chain = prompt | model | parser
```

```python
result = chain.invoke({"topic": "bears"})
print(result)
```

```python
#async strwaming meessage
async def run_ainvoke():
    result = await chain.ainvoke({"topic": "bears"})
    print(result)

await run_ainvoke()
```

```python
#batch
results = chain.batch(
    [
        {"topic": "bears"},
        {"topic": "lions"},
        {"topic": "wolves"},
    ]
)

print(results)
```

```python
#running abatch
async def run_abatch():
    results = await chain.abatch(
        [
            {"topic": "bears"},
            {"topic": "lions"},
            {"topic": "wolves"},
        ]
    )
    print(results)

await run_abatch()
```

```python
for chunk in chain.stream({"topic": "bears"}):
    print(chunk, end="", flush=True)
```

```python
async def run_astream():
    async for chunk in chain.astream({"topic":"bears"}):
        print(chunk, end="", flush=True)
await run_astream()
```

```python
from langchain_core.runnables import RunnableLambda
import random

def add_one(x: int) -> int:
  return x + 1

def buggy_double(y: int) -> int:
  """Buggy code that will fail 70% of the time"""
  if random.random() > 0.3:
      print('This code failed, and will probably be retried!')
      raise ValueError('Triggered buggy code')
  return y * 2

sequence = (
  RunnableLambda(add_one) |
  RunnableLambda(buggy_double).with_retry(
      stop_after_attempt=10,
      wait_exponential_jitter=False
  )
)

print(sequence.input_schema.model_json_schema())
print(sequence.output_schema.model_json_schema())
print(sequence.invoke(2))
```
